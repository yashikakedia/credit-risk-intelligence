```python
df = pd.read_csv(
    "../data/raw/accepted_2007_to_2018Q4.csv",
    low_memory=False
)

print("Dataset shape:", df.shape)
df.head()
```

```python
df["loan_status"].value_counts()
```

```python
bad_statuses = [
    "Charged Off",
    "Default",
    "Does not meet the credit policy. Status:Charged Off"
]

good_statuses = [
    "Fully Paid",
    "Does not meet the credit policy. Status:Fully Paid"
]

df_model = df[df["loan_status"].isin(bad_statuses + good_statuses)].copy()

df_model["target"] = df_model["loan_status"].isin(bad_statuses).astype(int)

print("Final modeling dataset shape:", df_model.shape)

print("\nTarget count:")
print(df_model["target"].value_counts())

print("\nTarget percentage:")
print(df_model["target"].value_counts(normalize=True).mul(100).round(2))
```

### Target Variable Definition

The target variable was created from the `loan_status` column. Loans with `Fully Paid` and `Does not meet the credit policy. Status:Fully Paid` were labelled as **0**, representing non-default borrowers.

Loans with `Charged Off`, `Default`, and `Does not meet the credit policy. Status:Charged Off` were labelled as **1**, representing default borrowers.

Active or uncertain loans such as `Current`, `Late`, and `In Grace Period` were excluded because their final repayment outcome was not confirmed.

After filtering, the final modeling dataset contains **1,348,099 records** and **152 columns**. The target distribution is **80.02% non-default borrowers** and **19.98% default borrowers**.

```python
dict_df = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.values.astype(str),
    "Non_Null_Count": df.notnull().sum().values,
    "Null_Percent": (df.isnull().mean() * 100).round(2).values,
    "Unique_Values": df.nunique().values,
    "Sample_Value": [
        str(df[c].dropna().iloc[0]) if df[c].notnull().any() else "ALL NULL"
        for c in df.columns
    ],
    "Description": "",
    "ML_Importance": ""
})

dict_df.to_excel("../report/data_dictionary.xlsx", index=False)

print("Data dictionary created successfully.")
```

### Data Dictionary Note

The data dictionary was created from the original raw LendingClub dataset (`df`) to document all available columns, data types, missing value percentages, unique values, and sample values. The filtered modeling dataset (`df_model`) was created separately only for defining the target variable and future modeling work.
